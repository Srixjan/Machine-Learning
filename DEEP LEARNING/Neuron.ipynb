```python
import torch
import torch.nn as nn
```

```python
# Input variables
torch.manual_seed(42) # lock randomizer.
inputs = torch.tensor([1.0, 2.0, 3.0])

# 0d -> Scalar(Constants)
# 1d -> Vector
# 2d -> Matrix
# 3d -> Tensors
```

```python
neuron = nn.Linear(in_features=3, out_features=1)
```

```python
output = neuron(inputs)
```
