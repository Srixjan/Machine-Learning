```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
```

```python
thyroid = pd.read_csv("thyroid_dataset.csv")
```

```python
thyroid.head(5)
```

```python
X = thyroid.drop("Outlier_label", axis=1)
y = thyroid["Outlier_label"]
```

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
X_scaled
```

```python
from sklearn.ensemble import IsolationForest

ilf = IsolationForest(
    n_estimators=300,
    contamination="auto",
    random_state=42
)

labels = ilf.fit_predict(X_scaled)
```

```python
# Visualisation
from sklearn.decomposition import PCA
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)
```

```python
plt.figure(figsize=(8, 6))
plt.scatter(x=X_pca[:, 0], y=X_pca[:, 1], c=labels)
plt.xlabel("PC1")
plt.ylabel("PCA2")
plt.show()

#Looks weird because we compressed the 22d to 2d using pca, but works good!
# use this for large dataset, and for small use DBSCAN.
```

```python
import numpy as np
n_outliers = np.sum(labels == -1)
n_normal = np.sum(labels == 1)

print("outlier: ", n_outliers)
print("normal values: ", n_normal)
```

```python
# LOF
from sklearn.neighbors import LocalOutlierFactor

lof = LocalOutlierFactor(contamination=0.036)

labels = lof.fit_predict(X_scaled)

plt.figure(figsize=(8, 6))
plt.scatter(x=X_pca[:, 0], y=X_pca[:, 1], c=labels)
plt.xlabel("PC1")
plt.ylabel("PCA2")
plt.show()
```
